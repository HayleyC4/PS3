## Utility: FASTA Parser

This function is used by all four problems. Complete it first.

A FASTA file looks like this:

```
>seq_001 | Site_A | Soil_sample_1
ATGCTTACGGATCGATCG...
TAGCTAGCTAGCGATCGA...
>seq_002 | Site_A | Soil_sample_2
ATGCTTACGGATCGATCG...
```

Each sequence can span multiple lines. Your parser needs to assemble those lines into a single string per record.

````python
def parse_fasta(filepath):
    """
    Reads a FASTA file and returns a list of (header, sequence) tuples.

    Parameters
    ----------
    filepath : str
        Path to the FASTA file.

    Returns
    -------
    list of tuple
        A list of (header, sequence) tuples, e.g.:
        [("seq_001 | Site_A | Soil_sample_1", "ATGC..."), ...]

    Hints
    -----
    - Use a for loop to read the file line by line.
    - Use an if statement to check whether a line starts with '>'.
    - You will need variables to keep track of the current header
      and the sequence being assembled across multiple lines.
    - What should you do when you encounter a new header while
      you already have a sequence stored?
    """

    records = []

    header = None
    sequence_parts = []

    with open(filepath, "r", encoding="utf-8") as fasta_file:
        for raw_line in fasta_file:
            line = raw_line.strip()

            if not line:
                continue

            if line.startswith(">"):
                if header is not None:
                    records.append((header, "".join(sequence_parts)))

                header = line[1:]
                sequence_parts = []
            else:
                if header is None:
                    raise ValueError("FASTA sequence data found before the first header")
                sequence_parts.append(line)

    if header is not None:
        records.append((header, "".join(sequence_parts)))

    return records


# Test your parser — this should print the header and first 40 characters
# of each sequence so you can verify it's working before moving on.
# Ensure FASTA_FILE exists even if the setup cell has not been run yet.
if "FASTA_FILE" not in globals():
    FASTA_FILE = "/workspaces/PS3/sequences.fasta"

records = parse_fasta(FASTA_FILE)
for header, seq in records:
    print(f"{header} : {seq[:40]}...")
````

## Problem 2: GC Content (5 pts)

Calculate and report GC content statistics across all sequences.

GC content is the fraction of bases that are G or C:

$$\text{GC}\% = \frac{\text{count}(G) + \text{count}(C)}{\text{total length}} \times 100$$

You must write a function `calculate_gc(sequence)` and call it from `report_gc_stats()`.

````python
def calculate_gc(sequence):
    """
    
    Calculates the GC content of a nucleotide sequence as a percentage.

    Parameters
    ----------
    sequence : str
        A nucleotide sequence string.

    Returns
    -------
    float
        GC content as a percentage (0.0 to 100.0).

    Hints
    -----
    - str.count() counts occurrences of a character.
    - Think about what to return if the sequence is empty.
    """
def calculate_gc(sequence):
    """Return the percentage of bases that are G or C."""
    sequence = sequence.upper()

    if not sequence:
        return 0.0

    gc_count = sequence.count("G") + sequence.count("C")
    return gc_count / len(sequence) * 100

    pass
````

````python
def report_gc_stats(records):
    """
    Prints GC content for each sequence and reports the highest,
    lowest, and average GC content across all sequences.

    Parameters
    ----------
    records : list of tuple
        A list of (header, sequence) tuples as returned by parse_fasta().

    Hints
    -----
    - Call calculate_gc() for each sequence.
    - Keep track of the highest and lowest values as you iterate.
    """

    # --- Your code here ---def report_gc_stats(records):
    if not records:
        print("No sequences to report.")
        return

    gc_percentages = []
    total_gc_bases = 0
    total_bases = 0

    for header, sequence in records:
        gc_percent = calculate_gc(sequence)
        gc_percentages.append(gc_percent)
        print(f"{header}: {gc_percent:.2f}% GC")

        total_gc_bases += sequence.upper().count("G") + sequence.upper().count("C")
        total_bases += len(sequence)

    print(f"\nMean per-sequence GC: {sum(gc_percentages) / len(gc_percentages):.2f}%")
    print(f"Lowest GC: {min(gc_percentages):.2f}%")
    print(f"Highest GC: {max(gc_percentages):.2f}%")

    if total_bases:
        overall_gc = total_gc_bases / total_bases * 100
        print(f"Overall GC across all bases: {overall_gc:.2f}%")

    pass


report_gc_stats(records)
````

**Bonus (+2 pts):** Make the motif a command-line argument so the script can be reused for any motif.

````python
def count_motif(sequence, motif):
    """
    Counts non-overlapping occurrences of a motif in a sequence.

    Parameters
    ----------
    sequence : str
        A nucleotide sequence string.
    motif : str
        The motif to search for (e.g., "GAATTC").

    Returns
    -------
    int
        Number of non-overlapping occurrences.

    Hints
    -----
    - Python strings have a built-in method that does exactly this.
    - Make the search case-insensitive.
    """

    # --- Your code here ---def count_motif(sequence, motif):
    sequence = sequence.upper()
    motif = motif.upper()

    if not motif:
        raise ValueError("Motif cannot be empty")

    count = 0
    start = 0

    while True:
        index = sequence.find(motif, start)
        if index == -1:
            return count
        count += 1
        start = index + 1  # Advance by one so overlapping matches count

    pass
````

````python
def report_motif_stats(records, motif="GAATTC"):
    """
    Searches for a motif in each sequence and prints a summary.

    Parameters
    ----------
    records : list of tuple
        A list of (header, sequence) tuples as returned by parse_fasta().
    motif : str, optional
        The motif to search for. Defaults to "GAATTC" (EcoRI site).
    """

def report_motif_stats(records, motif="GAATTC"):
    if not records:
        print("No sequences to search.")
        return

    counts = [(header, count_motif(sequence, motif))
              for header, sequence in records]

    for header, count in counts:
        print(f"{header}: {count}")

    highest_count = max(count for _, count in counts)
    most_sequences = [header for header, count in counts
                      if count == highest_count]
    print(f"Sequence(s) with the most occurrences ({highest_count}): "
          f"{', '.join(most_sequences)}")
    print(f"Total occurrences: {sum(count for _, count in counts)}")


def count_motif(sequence, motif):
    sequence = sequence.upper()
    motif = motif.upper()

    if not motif:
        raise ValueError("Motif cannot be empty")

    return sequence.count(motif)

pass


report_motif_stats(records, motif="GAATTC")
````

Warn if:
- The sequence contains characters other than `A`, `T`, `G`, `C`, or `N`


````python
def qc_check(records):
    """
    Quality control checker. Prints [PASS] or [WARN] for each sequence.

    Parameters
    ----------
    records : list of tuple
        A list of (header, sequence) tuples as returned by parse_fasta().

    Hints
    -----
    - Use a set of valid characters for fast membership checking.
    - A sequence with no data will be an empty string.
    - A single sequence can trigger more than one warning.
    """

    valid_bases = set("ATGCN")

def report_sequence_quality(records):
    valid_bases = set("ATGCN")

    for header, sequence in records:
        sequence = sequence.upper()
        issues = []

        if not sequence:
            issues.append("record has no sequence data")

        invalid_bases = sorted(set(sequence) - valid_bases)
        if invalid_bases:
            issues.append(f"invalid characters: {''.join(invalid_bases)}")

        if len(sequence) < 800:
            issues.append(f"shorter than 800 bp ({len(sequence)} bp)")
        elif len(sequence) > 1600:
            issues.append(f"longer than 1600 bp ({len(sequence)} bp)")

        if issues:
            print(f"[WARN] {header}: {'; '.join(issues)}")
        else:
            print(f"[PASS] {header}")


report_sequence_quality(records)

pass


qc_check(records)
````
